```python
summary = df.groupby('Class').agg(
    Total_Transactions=('Class', 'count'),
    Average_Amount=('Amount', 'mean'),
    Total_Amount=('Amount', 'sum')
).reset_index()

summary['Transaction_Type'] = summary['Class'].map({
    0: 'Genuine',
    1: 'Fraud'
})

summary['Percentage'] = (
    summary['Total_Transactions'] / len(df) * 100
)

summary = summary.round(2)

display(summary)
```
